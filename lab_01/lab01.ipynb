#### **Bài 4. Slicing và Indexing**
Truy xuất một phần dữ liệu từ Tensor nhiều chiều.

```python
a = torch.rand(3,4,5)
print("Original tensor:")
print(a)

print("\nTensor a[1]:")
print(a[1])
print("\nTensor a[1:, 2:4]:")
print(a[1:, 2:4])
```

**Yêu cầu:**

- a[1]: Lấy phần tử ở index 1 của chiều thứ nhất, bỏ đi chiều đó $\rightarrow$ Shape là (4, 5).
- a[1:, 2:4]: Lấy từ index 1 đến hết ở chiều 1 (còn lại 2 phần tử), lấy từ index 2 đến trước index 4 ở chiều 2 (còn lại $4 - 2 = 2$ phần tử), và giữ nguyên toàn bộ chiều 3 ($\rightarrow$ 5 phần tử) $\rightarrow$ Shape là (2, 2, 5).

• Thử lấy hai phần tử cuối cùng trên chiều thứ ba.

```python
# Chiều thứ ba là chiều cuối cùng (index -2 đến hết)
print(a[..., -2:])
```

#### **Bài 6. Phép toán và nhân ma trận**
Thực hiện các phép toán phần tử và phép nhân ma trận bằng toán tử @.

```python
a = torch.randn(2, 3)
b = torch.randn(2, 3)
print("a:")
print(a)
print("b:")
print(b)
print("\na + b:")
print(a + b)
print("\na / b:")
print(a / b)
print("\na^2:")
print(a ** 2)
print("\nMatrix multiplication:")
print(a @ b.T)
```

**Yêu cầu:**
•Phân biệt phép toán `a * b` với `a @ b.T.`:
- `a * b` (nhân từng phần tử - element-wise multiplication): Thực hiện nhân các phần tử có cùng vị trí với nhau. Yêu cầu a và b phải có cùng kích thước (shape).

- `a @ b.T` (phân tích nhân ma trận - matrix multiplication): Thực hiện nhân ma trận giữa a và chuyển vị của b (b.T). Kích thước cột của ma trận thứ nhất phải khớp với hàng của ma trận thứ hai.

**Cho biết shape của kết quả `a @ b.T:`**
- `a` có shape là` (2, 3)`.
- `b `có shape là `(2, 3) `$\rightarrow$ b.T (chuyển vị) có shape là `(3, 2).`
- Khi nhân ma trận `(2, 3) @ (3, 2)`, ta được kết quả có shape là `(2, 2).`

#### **Bài 8. Autograd cơ bản**
Tạo biến có `requires_grad=True` và dùng `backward()` để tính đạo hàm tự động.

```python
x = torch.tensor(
3.6,
requires_grad=True
)
y = x * x
y.backward()
print("x =", x)
print("y =", y)
print("x.grad =", x.grad)
```

**Yêu cầu:**
**Tính thủ công đạo hàm của $y = x^2$ tại $x = 3.6$ và so sánh với `x.grad`:**
- Công thức đạo hàm: $\frac{dy}{dx} = 2x$.
- Tại $x = 3.6$: Đạo hàm bằng $2 \times 3.6 = 7.2$.
- So sánh: Giá trị x.grad in ra màn hình cũng sẽ bằng 7.2 (có thể lệch đôi chút do kiểu dữ liệu số thực trôi floating-point).

•**Đổi hàm thành $y = 3x^2 + 2x + 1$ và kiểm tra gradient:**


```python
x = torch.tensor(3.6, requires_grad=True)
y = 3 * x**2 + 2 * x + 1
y.backward()
print("x.grad mới =", x.grad)
# Đạo hàm thủ công: 6x + 2 -> 6(3.6) + 2 = 21.6 + 2 = 23.6
```

#### **Bài 9. Gradient Descent tìm hệ số đa thức**
Sinh dữ liệu từ đa thức y = x2 + 2x + 3, sau đó dùng NAdam và autograd để tìm lại ba hệ số.

```python
import numpy as np
import torch
polynomial = np.poly1d([1, 2, 3])
N = 20
X = np.random.randn(N, 1) * 5
Y = polynomial(X)
XX = np.hstack([
    X * X,
    X,
    np.ones_like(X)
])
w = torch.randn(3, 1, requires_grad=True)
x = torch.tensor(XX, dtype=torch.float32)
y = torch.tensor(Y, dtype=torch.float32)
optimizer = torch.optim.NAdam([w], lr=0.01)

print("Initial coefficients:")
print(w)
for epoch in range(1000):
    optimizer.zero_grad()
    y_pred = x @ w
    mse = torch.mean(torch.square(y - y_pred))
    mse.backward()
    optimizer.step()
print("Learned coefficients:")
print(w)
```

**Yêu cầu:**
• So sánh ba hệ số học được với` [1, 2, 3]`.
- **Hệ số ban đầu (Initial coefficients):**` tensor([[0.6864], [-0.8580], [1.2355]])` (được khởi tạo ngẫu nhiên nên rất xa giá trị chuẩn).
- **Hệ số học được (Learned coefficients):** `tensor([[1.0011], [2.0023], [2.8964]]).`
- **Nhận xét**:** Ba hệ số học được tiến rất sát với bộ hệ số gốc `[1, 2, 3]` của đa thức: 
    - Hệ số bậc 2: $1.0011 \approx 1$
    - Hệ số bậc 1: $2.0023 \approx 2$
    - Hệ số tự do: $2.8964 \approx 3$ (sai số rất nhỏ, do thuật toán Gradient Descent tối ưu qua 1000 vòng lặp).

**• Giải thích vai trò của optimizer.zero_grad(), mse.backward() và optimizer.step().**
- `optimizer.zero_grad()`: Xóa sạch (đặt về 0) các gradient tích lũy từ vòng lặp trước đó để tránh việc gradient bị cộng dồn nhầm.
- `mse.backward():` Kích hoạt cơ chế tự động tính đạo hàm (autograd) lan truyền ngược qua đồ thị tính toán để tính gradient của hàm lỗi (MSE) theo các hệ số `w`.
- `optimizer.step():` Sử dụng thuật toán tối ưu (NAdam) để cập nhật giá trị mới cho các hệ số w dựa trên các gradient vừa tính nhằm làm giảm sai số.

#### **Bài 10. Giải hệ phương trình bằng Autograd**
Tối thiểu hóa tổng sai số bình phương để tìm A, B, C, D thỏa bốn phương trình.

```python
import random
import torch
A = torch.tensor(random.random(), requires_grad=True)
B = torch.tensor(random.random(), requires_grad=True)
C = torch.tensor(random.random(), requires_grad=True)
D = torch.tensor(random.random(), requires_grad=True)

EPOCHS = 2000
optimizer = torch.optim.NAdam([A, B, C, D], lr=0.01)

for _ in range(EPOCHS):
    y1 = A + B - 9
    y2 = C - D - 1
    y3 = A + C - 8
    y4 = B - D - 2

    sqerr = y1*y1 + y2*y2 + y3*y3 + y4*y4
    optimizer.zero_grad()
    sqerr.backward()
    optimizer.step()

print("A =", A)
print("B =", B)
print("C =", C)
print("D =", D)
```

**Yêu cầu:**

- Kiểm tra nghiệm thu được với bốn phương trình ($A+B=9, C-D=1, A+C=8, B-D=2$):
    - Sau khi chạy thuật toán tối ưu hóa, các giá trị $A, B, C, D$ tìm được sẽ thỏa mãn xấp xỉ hệ phương trình đề bài cho (tổng bình phương sai số sqerr giảm về gần bằng 0).

- **Nhận xét vì sao bài toán có thể có nhiều nghiệm.**
    - Hệ phương trình đề bài cho gồm 4 phương trình nhưng thực chất chỉ có 3 phương trình độc lập tuyến tính (vì nếu lấy phương trình (1) trừ phương trình (4) vế theo vế, ta sẽ thu được đúng phương trình (3): $(A+B) - (B-D) = A + D = 9 - 2 = 7$, hoặc sự liên kết giữa các ẩn tạo thành một hệ có vô số nghiệm phụ thuộc vào một tham số tự do).
    - Vì số ẩn ($4$ ẩn: $A, B, C, D$) nhiều hơn số ràng buộc độc lập thực sự, hệ có vô số nghiệm. Do đó, thuật toán tối ưu sẽ hội tụ dừng lại tại một nghiệm cụ thể phụ thuộc vào điểm khởi tạo ngẫu nhiên ban đầu `(random.random())`.

**4. Huấn luyện lại Bài 9 với đa thức $y = 2x^2 - 3x + 5$**

```python
import numpy as np

# Đa thức mới: y = 2x² - 3x + 5
polynomial = np.poly1d([2, -3, 5])
N = 20

# Sinh dữ liệu
X = np.random.randn(N, 1) * 5
Y = polynomial(X)

# Tạo ma trận đặc trưng [x², x, 1]
XX = np.hstack([
    X * X,
    X,
    np.ones_like(X)
])

# Khởi tạo 3 hệ số ngẫu nhiên
w = torch.randn(3, 1, requires_grad=True)

x = torch.tensor(XX, dtype=torch.float32)
y = torch.tensor(Y, dtype=torch.float32)

# Optimizer
optimizer = torch.optim.NAdam([w], lr=0.01)

print("Initial coefficients:")
print(w)

# Huấn luyện
for epoch in range(1000):
    optimizer.zero_grad()

    # y_pred = a*x² + b*x + c
    y_pred = x @ w

    # MSE
    mse = torch.mean(torch.square(y - y_pred))

    # Tính gradient
    mse.backward()

    # Cập nhật hệ số
    optimizer.step()

print("Learned coefficients:")
print(w)
```
