#### __Bài 1. Tạo Tensor cơ bản__

Tạo một Tensor số nguyên kích thước 2 × 3 và hiển thị kết quả.

```python
import torch
a = torch.tensor(
    [[1, 2, 3],
     [4, 5, 6]],
     dtype=torch.int32
)
print(a)
```

**Yêu cầu**: Cho biết kích thước, số chiều và kiểu dữ liệu của Tensor a.

```python
# kich thuoc cuar tensor
print("kich thuoc:", a.shape)
# so chieu cua tensor
print("So chieu:", a.dim())
# kieu du lieu cua tensor
print("kieu du lieu:",a.dtype)
```

#### **Bài 2. Tạo các Tensor thường dùng**
Thực hành tạo Tensor ngẫu nhiên, Tensor toàn số 0, toàn số 1 và ma trận đơn vị.

```python
a = torch.rand(3, 4)
b = torch.randn(3, 4)
c = torch.zeros(2, 3)
d = torch.ones(2, 3)
e = torch.eye(4)

print(a)
print(b)
print(c)
print(d)
print(e)
```

**Yêu cầu:**

• Nhận xét sự khác nhau giữa torch.rand() và torch.randn():

`torch.rand()`: Tạo tensor với các giá trị ngẫu nhiên tuân theo phân phối đều (uniform distribution) trong khoảng từ [0, 1).

`torch.randn()`:Tạo tensor với các giá trị ngẫu nhiên tuân theo phân phối chuẩn/chuông (normal/gaussian distribution) với trung bình bằng 0 và độ lệch chuẩn bằng 1 (các giá trị có thể âm hoặc lớn hơn 1).

• **Tạo thêm một Tensor 5 × 5 chứa toàn giá trị 10.**

```python
tensor_10 = torch.ones(5, 5)* 10
print(tensor_10)
```

#### **Bài 3. Kiểm tra thông tin Tensor**
Sử dụng shape, size(), ndim và dtype để kiểm tra các thuộc tính của Tensor.


```python
a = torch.zeros(2,3, 4)
print("Tensor:")
print(a)
print("Shape:", a.shape)
print("Size:", a.size())
print("Number of dimensions:", a.ndim)
print("Data type:", a.dtype)
```

**Yêu cầu:** Giải thích ý nghĩa của torch.Size([2, 3, 4])
- Tensor này có tổng cộng 3 chiều (3 cấp độ ngoặc vuông).
- Chiều thứ nhất: Có kích thước là 2 (chứa 2 khối dữ liệu con bên trong).
- Chiều thứ hai (ở giữa): Mỗi khối con có 3 hàng
- Chiều thứ ba : Mỗi hàng có 4 cột (4 phần tử).
- Tổng số phần tử của tensor này là $2 \times 3 \times 4 = 24$ phần tử



#### **Bài 4. Slicing và Indexing**
Truy xuất một phần dữ liệu từ Tensor nhiều chiều.

```python
a = torch.rand(3,4,5)
print("Original tensor:")
print(a)

print("\nTensor a[1]:")
print(a[1])
print("\nTensor a[1:, 2:4]:")
print(a[1:, 2:4])
```

**Yêu cầu:**

• Cho biết shape của a[1] và a[1:, 2:4].

- a ban đầu có shape là (3, 4, 5).
- a[1]: Lấy phần tử ở index 1 của chiều thứ nhất, bỏ đi chiều đó $\rightarrow$ Shape là (4, 5).
- a[1:, 2:4]: Lấy từ index 1 đến hết ở chiều 1 (còn lại 2 phần tử), lấy từ index 2 đến trước index 4 ở chiều 2 (còn lại $4 - 2 = 2$ phần tử), và giữ nguyên toàn bộ chiều 3 ($\rightarrow$ 5 phần tử) $\rightarrow$ Shape là (2, 2, 5).

• Thử lấy hai phần tử cuối cùng trên chiều thứ ba.

```python
# Chiều thứ ba là chiều cuối cùng (index -2 đến hết)
print(a[..., -2:])
```

### **Bài 5. Reshape và Flatten**
Thay đổi hình dạng Tensor mà không làm thay đổi số phần tử.

```python
a = torch.randn(3, 4)
print("Original: ")
print(a)
print("\nFlatten:")
print(a.ravel())

print("\nReshape to 3 x 2 x 2:")
print(a.reshape(3, 2, 2))
```

**Yêu cầu:**

• Kiểm tra số phần tử trước và sau khi reshape.

- số phần tử trước khi reshape: $3 \times 4 = 12$ phần tử.
- Số phần tử sau khi reshape (thành $3 \times 2 \times 2$ hoặc $2 \times 6$): $3 \times 2 \times 2 = 12$ và $2 \times 6 = 12$.
==> Số phần tử được bảo toàn hoàn toàn (đều là 12 phần tử), miễn là tích các chiều mới bằng tổng số phần tử ban đầu.

• Thử reshape Tensor thành kích thước 2 × 6.

```python
a = torch.randn(3, 4)
# Reshape thành ma trận 2 hàng, 6 cột
print(a.reshape(2, 6))
```
